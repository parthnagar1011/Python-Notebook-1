WAF to print the factorail of n.

```python
def fact(n) :
  fact = 1
  if (n == 0 or n == 1) :
    print("The Factorial is 1.")

  else :

    for i in range(1,n+1) :
      fact *= i
      print(fact)

fact(5)
```

## Recursion

Recursion is when a function calls itself to solve smaller instances of the same problem.

We need to stop the recursive case using base case so it does not run infinite time.


```python
# Example - Print 1 to n backwards

def show(n) :

  if (n == 0) :
    return

  print(n)
  show(n-1)

show(5)
```

Factorial program using recursion.

```python
def fact(n) :
  if(n == 0 or n == 1) :
    return 1

  else :
    return n * fact(n-1)

print(fact(5))
```

WRF to calculate sum of first n natural numbers.

```python
def nsum(n) :
  if(n == 0) :
    return 0
  return nsum(n-1) + n

sum = nsum(5)
print(sum)
```

## FILE IO

File I/O stands for File Input/Output. It means reading data from or writing data to a file.

Python allows us to open, read, write, and close files using simple commands.

Modes:

"r" – Read (default mode)

"w" – Write (creates file if not exists)

"a" – Append (adds to existing content)

"x" – Create (fails if file exists)

"b" – Binary mode

"t" – Text mode (default)



```python
file_x = open("demo.txt" , "r")
data = file_x.read()
print(data)
file_x.close()
```
